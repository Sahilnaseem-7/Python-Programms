```python
#Checking a number if its positive of negative
num = int(input("Enter the number to check : "))
if num > 0:
    print(f"The given number {num} is positive number")
elif num == 0:
    print(f"The given number {num} is zero ")
else :
    print(f"The given number {num} is negative number")
```

```python
#Checking a number if its odd or even 
num = int(input("Enter the given number to check "))
if num%2 == 0:
    print(f"The given number {num} is even number")
else :
    print(f"The given number {num} is odd")
```

```python
# Checking if its a leap year
year = int(input("Enter the year to check"))

#If divided by 100 it means its a century year
#if divided by 400 it means its a century leap yaer
if (year%400 == 0) and (year%100 == 0):
    print(f"The given year {year} is a century leap year")

#Not divided by 100 it means its not a century year
#Diidec by 4 means it a leap year
elif (year%4 ==0) and (year%100 != 0):
    print(f"The given year {year} is a leap year")

else:
    print(f"The year {year} is not a leap year")
```

```python
#Program to check for prime numbers 
num = int(input("Enter a number: "))

if num < 2:
    print(f"The number {num} is Not prime")
else:
    #Checking n the for loop
    for i in range(2, num):
        if num % i == 0:
            #If number is divided by any of the number in the range (2,num) then it is not prime as it should be divided by itself and 1
            print(f"The number {num} is Not prime")
            break
    else:
        print(f"The number {num} is Prime")
```
