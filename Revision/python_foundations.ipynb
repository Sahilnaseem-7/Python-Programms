```python
#Airthmetics
num1 = float(input("Enter the desired number for performing airthemtics \n"))
num2 = float(input("Enter the second number\n"))
addition = num1 + num2
subtraction = num1 - num2 
multiplication = num1 * num2
if num2==0:
    print("Divison by zero not allowed")
else:
    division = num2/num1    
print(f"Addition = {num1} + {num2} = {addition}\n"
      f"Substraction = {num1} - {num2} = {subtraction}\n"
      f"Multiplication = {num1} * {num2} = {multiplication}\n"
      f"Divison = {num2} / {num1} = {division}")

```

```python
#Area of Triangle
height = float(input("Enter the height of the triangle\n"))
base = float(input("Enter the base of the triangle\n"))
area = 0.5 * base * height
print(f"The area of triangle with base {base} and height {height} is {area}")
```

```python
#Swapping Variables 
a = input("Enter the value of the first variable\n")
b = input("Enter the value of the second variable \n")
print(f"Original variables A = {a} B = {b}")
#Swapp the values using a temporary variable(Temporary assingment)
temp = a
a = b
b = temp
print(f"Swapped variables : A = {a} B = {b} ")

```

```python
#Generating random number
import random
print(f"Random numbers between 1 to 1000 : {random.randint(1,1000)}")
#adding randoms 
num1 = random.randint(1, 100)
num2 = random.randint(1,1000)
addition = num1 + num2
print(f"Addition of random numbers {num1} and {num2} = {addition}")
#Multiplicatio
num3 = random.randint(1,100)
num4 = random.randint(1,100)
multiplication = num3 * num4
print(f"Muliplication of random numbers {num3} and {num4} = {multiplication}")
```
